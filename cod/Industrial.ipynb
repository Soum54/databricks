```python
%sql
use catalog cat_ltm10877547;
use schema schema_data_integration;
show schemas in cat_ltm10877547;
```

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
schema=StructType([
 StructField("asset_id",StringType(),True),
 StructField("region",StringType(),True),
 StructField("asset_type",StringType(),True),
 StructField("quantity",IntegerType(),True),
 StructField("unit_cost",DecimalType(12,2),True),
 StructField("total_cost",DecimalType(14,2),True),
 StructField("condition",StringType(),True)
])
df=spark.read.csv("/Volumes/cat_ltm10877547/schema_data_integration/ext_loc_soumiya/assessments/10877547/m01/M01_Industrial_Asset_Delta_Maintenance.csv",header=True,schema=schema)
```

```python
df.write.format("delta").mode("overwrite").option("overwriteSchema", "true").saveAsTable("cat_ltm10877547.schema_data_integration.asset_costs")
```
